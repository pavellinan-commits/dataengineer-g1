### Pregunta 1.3: Contar frecuencia de elementos

Dada una lista de nombres repetidos, muestra **cuántas veces aparece cada nombre**.
Cada nombre debe mostrarse una sola vez, en el orden en que aparece por primera vez.

```python
nombres = ['Ana', 'Luis', 'Ana', 'Carlos', 'Luis', 'Ana', 'María', 'Carlos']
```

```python
# Escribe tu solución aquí
frecuencias = {}
for nombre in nombres:
    if nombre not in frecuencias:
        frecuencias[nombre] = 0
    frecuencias[nombre] += 1

for nom, cant in frecuencias.items():
    print(f"{nom} aparece {cant} veces")
```

### Pregunta 2.4: Estadísticas de una tupla

Con la tupla `notas`, calcula el **promedio**, la **nota máxima**, la **nota mínima** y la
**cantidad de aprobados** (nota mayor o igual a 11).

```python
notas = (18, 15, 20, 11, 14, 17, 19)
```

```python
# Escribe tu solución aquí
total = 0
maximo = notas[0]
minimo = notas[0]
aprobados = 0

for nota in notas:
    total += nota
    if nota > maximo:
        maximo = nota
    if nota < minimo:
        minimo = nota
    if nota >= 11:
        aprobados += 1

promedio = total / len(notas)
print(f"Promedio: {promedio:.2f}")
print(f"Nota máxima: {maximo}")
print(f"Nota mínima: {minimo}")
print(f"Aprobados: {aprobados}")
```

### Pregunta 3.1: Consolidación de inventarios

Combina tres inventarios provenientes de distintas sucursales. Cuando un producto aparezca en varias
sucursales, **suma sus cantidades** y calcula también **en cuántas sucursales** está disponible.

**Requisitos:**
- Crea un diccionario final donde cada producto tenga las claves `cantidad_total` y `sucursales`.
- No modifiques los diccionarios originales.
- Ordena el resultado alfabéticamente por producto.

```python
inventario_norte = {"laptop": 4, "mouse": 10, "monitor": 3}
inventario_sur = {"laptop": 2, "teclado": 8, "monitor": 5}
inventario_centro = {"mouse": 6, "teclado": 4, "webcam": 7}
```

```python
# Escribe tu solución aquí
inventario_total = {}

inventarios = [inventario_norte, inventario_sur, inventario_centro]

articulos = set([
    *inventario_norte.keys(),
    *inventario_sur.keys(),
    *inventario_centro.keys(),
])

print(articulos)
print("="*50)

for articulo in articulos:
    inventario_total[articulo] = {
        "cantidad_total": sum([inventario.get(articulo, 0) for inventario in inventarios]),
        "sucursales": sum([1 for inventario in inventarios if articulo in inventario])
    }

for articulo in inventario_total:
    print(f"articulo: {articulo} | {inventario_total[articulo]}")
    print('='*50)
```

### Pregunta 3.3: Agrupación jerárquica de empleados

Agrupa a los empleados primero por **departamento** y luego por **nivel** de experiencia.

**Requisitos:**
- En cada grupo almacena `nombres`, `cantidad` y `salario_promedio`.
- La estructura debe ser un diccionario anidado: `{departamento: {nivel: {...}}}`.

```python
empleados = [
    {"nombre": "Ana", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5200},
    {"nombre": "Luis", "departamento": "Diseño", "nivel": "Junior", "salario": 2800},
    {"nombre": "Carla", "departamento": "Desarrollo", "nivel": "Junior", "salario": 3400},
    {"nombre": "Pedro", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5600},
    {"nombre": "Lucía", "departamento": "Diseño", "nivel": "Senior", "salario": 4700},
]
```

```python
# Escribe tu solución aquí
grupos = {}

for emp in empleados:
    dep = emp["departamento"]
    niv = emp["nivel"]
    
    if dep not in grupos:
        grupos[dep] = {}
        
    if niv not in grupos[dep]:
        grupos[dep][niv] = {"nombres": [], "salarios": []}
        
    grupos[dep][niv]["nombres"].append(emp["nombre"])
    grupos[dep][niv]["salarios"].append(emp["salario"])

for dep, niveles in grupos.items():
    for niv, datos in niveles.items():
        cant = len(datos["nombres"])
        promedio = sum(datos["salarios"]) / cant
        print(f"Depto: {dep} | Nivel: {niv}")
        print(f"  Nombres: {datos['nombres']}")
        print(f"  Cantidad: {cant}")
        print(f"  Salario Promedio: {promedio:.2f}")
```

### Pregunta 3.4: Sistema de votación con validación

Procesa una lista de votos y separa los votos **válidos** de los **inválidos**.

**Requisitos:**
- Los votos deben ser insensibles a mayúsculas (`"ANA"` cuenta para `"Ana"`).
- Registra el conteo por candidato, los votos inválidos y el porcentaje obtenido (sobre los votos válidos).
- Determina si hay un ganador o un empate.

```python
candidatos = {"Ana", "Luis", "Carla"}
votos = ["Ana", "Luis", "ana", "Carla", "Pedro", "Luis", "ANA", "", "Carla"]
```

```python
# Escribe tu solución aquí
votos_candidatos = {}
for c in candidatos:
    votos_candidatos[c.lower()] = 0
    
validos = 0
invalidos = 0

for v in votos:
    v_limpio = v.strip().lower()
    if v_limpio in votos_candidatos:
        votos_candidatos[v_limpio] += 1
        validos += 1
    else:
        invalidos += 1

print(f"Total válidos: {validos}")
print(f"Total inválidos: {invalidos}")

max_votos = 0
ganadores = []

for c, cantidad in votos_candidatos.items():
    if validos > 0:
        porcentaje = (cantidad / validos) * 100
    else:
        porcentaje = 0
    print(f"{c.title()}: {cantidad} votos ({porcentaje:.2f}%)")
    
    if cantidad > max_votos:
        max_votos = cantidad
        ganadores = [c.title()]
    elif cantidad == max_votos:
        ganadores.append(c.title())

if len(ganadores) == 1:
    print(f"Ganador: {ganadores[0]}")
elif len(ganadores) > 1:
    print(f"Empate entre: {ganadores}")
```

### Pregunta 3.5: Índice invertido de documentos

Crea un **índice invertido** que permita saber en qué documentos aparece cada palabra.

**Requisitos:**
- Cada palabra debe apuntar a un **conjunto** (`set`) de identificadores de documentos.
- Ignora las palabras de una sola letra.
- Usa el índice para responder: ¿en qué documentos aparecen `"python"` y `"datos"`?

```python
documentos = {
    "doc1": "python y ciencia de datos",
    "doc2": "python para desarrollo web",
    "doc3": "ciencia de datos con inteligencia artificial",
}
```

```python
# Escribe tu solución aquí
indice = {}

for doc_id, texto in documentos.items():
    palabras = texto.split()
    for p in palabras:
        if len(p) > 1:
            if p not in indice:
                indice[p] = set()
            indice[p].add(doc_id)

for p, docs in indice.items():
    print(f"{p}: {docs}")

python_docs = indice.get("python", set())
datos_docs = indice.get("datos", set())
ambos = python_docs.intersection(datos_docs)
print(f"Aparecen en ambos: {ambos}")
```

### Pregunta 5.1: Analizar las ventas de una tienda

Crea una función `analizar_ventas(ventas)` que reciba una lista de montos de venta y construya un resumen estadístico.

**Requisitos:**
- Calcular el total vendido, el promedio, la venta más alta y la venta más baja.
- Contar cuántas ventas fueron superiores al promedio.
- Indicar cuántas ventas se encuentran por debajo de 100.
- Retornar todos los resultados dentro de un diccionario.
- Si la lista está vacía o contiene valores negativos, retornar un mensaje de error.

**Reto adicional:** agrega un parámetro opcional `meta`. La función debe indicar si el total vendido alcanzó dicha meta.

```python
ventas = [150.50, 200, 80.90, 300, 175.50, 90]
```

```python
# Escribe tu solución aquí
def analizar_ventas(ventas, meta=None):
    if len(ventas) == 0:
        return "Error: la lista está vacía"
        
    for v in ventas:
        if v < 0:
            return "Error: hay ventas negativas"
            
    total = sum(ventas)
    promedio = total / len(ventas)
    maxima = max(ventas)
    minima = min(ventas)
    
    superan_prom = 0
    bajo_100 = 0
    
    for v in ventas:
        if v > promedio:
            superan_prom += 1
        if v < 100:
            bajo_100 += 1
            
    resultado = {
        "total": total,
        "promedio": promedio,
        "maxima": maxima,
        "minima": minima,
        "mas_que_promedio": superan_prom,
        "bajo_100": bajo_100
    }
    
    if meta is not None:
        resultado["alcanzo_meta"] = total >= meta
        
    return resultado

print(analizar_ventas(ventas, meta=1000))
```

```python
stats = analizar_ventas([])
print(stats)
```

```python
stats = analizar_ventas([150.50, 200, 80.90, 300, 175.50, 90, -100])
print(stats)
```

```python
# Escribe tu solución aquí
def validar_password(password):
    errores = []
    
    if len(password) < 8:
        errores.append("Muy corta, necesita 8 caracteres")
        
    tiene_mayus = False
    tiene_minus = False
    tiene_num = False
    
    for letra in password:
        if letra.isupper():
            tiene_mayus = True
        elif letra.islower():
            tiene_minus = True
        elif letra.isdigit():
            tiene_num = True
            
    if not tiene_mayus:
        errores.append("Falta mayúscula")
    if not tiene_minus:
        errores.append("Falta minúscula")
    if not tiene_num:
        errores.append("Falta un número")
        
    simbolos = ["@", "#", "$", "%"]
    tiene_simbolo = False
    for s in simbolos:
        if s in password:
            tiene_simbolo = True
            break
            
    if not tiene_simbolo:
        errores.append("Falta símbolo (@, #, $, %)")
        
    if " " in password:
        errores.append("No puede tener espacios")
        
    es_valida = len(errores) == 0
    return (es_valida, errores)

valida, fallos = validar_password(password_prueba)
if valida:
    print("Contraseña correcta")
else:
    print("Contraseña incorrecta por:")
    for f in fallos:
        print("-", f)
```

**Requisitos:**
- Cada empleado tendrá una lista de puntajes entre 0 y 100.
- Calcular el promedio de cada empleado.
- Clasificar el resultado como **Excelente** (≥ 90), **Bueno** (≥ 80), **Regular** (≥ 70) o **Deficiente** (< 70).
- Incluir únicamente a quienes tengan un promedio igual o superior a `minimo`.
- Retornar una lista de tuplas con `(nombre, promedio, clasificación)`.
- Ordenar de mayor a menor promedio y, en caso de empate, alfabéticamente.

```python
empleados = {
    'Ana': [80, 90, 75],
    'Carlos': [60, 70, 65],
    'Lucía': [95, 85, 90],
    'Pedro': [70, 70, 70],
}
```

```python
# Escribe tu solución aquí
def evaluar_empleados(empleados, minimo=70):
    resultados = []
    
    for nombre, notas in empleados.items():
        promedio = sum(notas) / len(notas)
        
        if promedio >= minimo:
            if promedio >= 90:
                clasificacion = "Excelente"
            elif promedio >= 80:
                clasificacion = "Bueno"
            elif promedio >= 70:
                clasificacion = "Regular"
            else:
                clasificacion = "Deficiente"
                
            resultados.append((nombre, promedio, clasificacion))
            
    # Ordenar primero por nombre, luego por promedio al reves
    resultados.sort(key=lambda x: x[0])
    resultados.sort(key=lambda x: x[1], reverse=True)
    
    return resultados

print(evaluar_empleados(empleados))
```

Implementa dos funciones: `comprimir_texto(texto)` y `descomprimir_texto(texto)`.

**Requisitos:**
- `comprimir_texto` debe reemplazar caracteres consecutivos repetidos por el carácter y su cantidad.
  Por ejemplo, `aaabbccccdaa` debe convertirse en `a3b2c4d1a2`.
- `descomprimir_texto` debe reconstruir el texto original.
- La descompresión debe admitir cantidades de más de un dígito, como `a12`.
- Si el formato comprimido es inválido, retornar un mensaje de error.

```python
texto = 'aaabbccccdaa'
```

```python
# Escribe tu solución aquí
def comprimir_texto(texto):
    if texto == "":
        return ""
        
    resultado = ""
    letra_actual = texto[0]
    contador = 0
    
    for letra in texto:
        if letra == letra_actual:
            contador += 1
        else:
            resultado += letra_actual + str(contador)
            letra_actual = letra
            contador = 1
            
    resultado += letra_actual + str(contador)
    return resultado

def descomprimir_texto(texto):
    resultado = ""
    letra_actual = ""
    numero_actual = ""
    
    for caracter in texto:
        if caracter.isalpha():
            if letra_actual != "":
                resultado += letra_actual * int(numero_actual)
            letra_actual = caracter
            numero_actual = ""
        else:
            numero_actual += caracter
            
    if letra_actual != "":
        resultado += letra_actual * int(numero_actual)
        
    return resultado

comp = comprimir_texto(texto)
print("Comprimido:", comp)
print("Descomprimido:", descomprimir_texto(comp))
```

- Realizar la reserva únicamente si todos están disponibles.
- Evitar reservas parciales y mantener la sala intacta si ocurre un error.
- Retornar el estado, los asientos reservados y la sala actualizada.

**Reto adicional:** crea otra función `mostrar_sala(sala)` que muestre filas y columnas de manera legible.

```python
sala = [['L', 'L', 'O', 'L'],
        ['O', 'L', 'L', 'L'],
        ['L', 'O', 'L', 'L']]
asientos = [(1, 2), (2, 3), (3, 1)]
```

```python
# Escribe tu solución aquí
def mostrar_sala(sala):
    print("   C1 C2 C3 C4")
    print("  ------------")
    num_fila = 1
    for fila in sala:
        fila_str = ""
        for asiento in fila:
            fila_str += asiento + "  "
        print(f"F{num_fila}| {fila_str}")
        num_fila += 1

def reservar_asientos(sala, asientos_solicitados):
    # Validar primero
    for (f, c) in asientos_solicitados:
        f_idx = f - 1
        c_idx = c - 1
        
        if f_idx < 0 or f_idx >= len(sala) or c_idx < 0 or c_idx >= len(sala[0]):
            return "Error: Asiento fuera de rango", [], sala
            
        if sala[f_idx][c_idx] == 'O':
            return "Error: Un asiento ya está ocupado", [], sala
            
    # Reservar
    for (f, c) in asientos_solicitados:
        sala[f-1][c-1] = 'O'
        
    return "Reserva exitosa", asientos_solicitados, sala

mostrar_sala(sala)
estado, reservados, nueva_sala = reservar_asientos(sala, asientos)
print(estado)
mostrar_sala(nueva_sala)
```

### Pregunta 6.1: Clase CuentaBancaria

- Método `depositar(monto)`: suma al saldo, solo si el monto es positivo.
- Método `retirar(monto)`: resta del saldo, pero debe impedir que el saldo quede negativo y avisar `"Fondos insuficientes"`.
- Método `consultar_saldo()`: devuelve el saldo actual.

Prueba la clase con una cuenta: deposita, retira un monto válido y luego intenta retirar más de lo disponible.

*Objetivo: modificar el estado interno del objeto con validaciones.*

```python
# No hay datos de entrada: crea tus propios objetos para probar la clase.
```

```python
# Escribe tu solución aquí
class CuentaBancaria:
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self.saldo = saldo_inicial
        
    def depositar(self, monto):
        if monto > 0:
            self.saldo += monto
            print(f"Depositado: {monto}")
        else:
            print("El monto debe ser positivo")
            
    def retirar(self, monto):
        if monto > self.saldo:
            print("Fondos insuficientes")
        else:
            self.saldo -= monto
            print(f"Retirado: {monto}")
            
    def consultar_saldo(self):
        return self.saldo

cuenta = CuentaBancaria("Jhon Snow", 100)
cuenta.depositar(200)
cuenta.retirar(50)
cuenta.retirar(10000)
print("Saldo final:", cuenta.consultar_saldo())
```

```python
# Escribe tu solución aquí
class Animal:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad
        
    def hacer_sonido(self):
        return "Sonido genérico"
        
class Perro(Animal):
    def __init__(self, nombre, edad, raza):
        super().__init__(nombre, edad)
        self.raza = raza
        
    def hacer_sonido(self):
        return "guau!"
        
class Gato(Animal):
    def hacer_sonido(self):
        return "Miau!"

mascotas = [
    Perro("Fido", 3, "Chusco"),
    Gato("Michifus", 2),
    Animal("Desconocido", 1)
]

for mascota in mascotas:
    print(mascota.nombre, "hace:", mascota.hacer_sonido())
```

```python
class Motor:
    def __init__(self, tipo, caballos):
        self.tipo = tipo
        self.caballos = caballos
        self.encendido = False

    def arrancar(self):
        if self.encendido:
            return "El motor ya estaba encendido."

        self.encendido = True
        return f"Motor {self.tipo} de {self.caballos} caballos encendido."


class Vehiculo:
    def __init__(self, marca, motor):
        self.marca = marca
        self.motor = motor

    def arrancar(self):
        return self.motor.arrancar()


motor = Motor("Dual", 220)
vehiculo = Vehiculo("Kia", motor)

print(vehiculo.arrancar())
print(vehiculo.arrancar())
```

### Pregunta 6.5: Sistema de biblioteca (ejercicio integrador)

Crea tres clases que trabajen juntas: `Libro`, `Usuario` y `Biblioteca`.

- `Libro`: `titulo`, `autor` y `disponible` (`True` por defecto). *(Es una clase distinta a la del ejercicio 6.2: puedes redefinirla.)*
- `Usuario`: `nombre` y una lista de `libros_prestados`.
- `Biblioteca`: guarda una lista de libros y ofrece los métodos `agregar_libro()`, `buscar(titulo)`,
  `prestar(titulo, usuario)`, `devolver(titulo, usuario)` y `catalogo()`.
- Al prestar un libro debe marcarse como no disponible y añadirse a la lista del usuario; al devolverlo, lo contrario.
- Contempla los casos: el libro no existe, el libro ya está prestado, el usuario no tiene ese libro.

*Objetivo: integrar varias clases que se relacionan entre sí, tal como ocurre en un programa real.*

```python
# No hay datos de entrada: crea tus propios objetos para probar las clases.
```

```python
# Escribe tu solución aquí
class Libro_Biblio:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor
        self.disponible = True

class Usuario:
    def __init__(self, nombre):
        self.nombre = nombre
        self.libros_prestados = []

class Biblioteca:
    def __init__(self):
        self.lista_libros = []
        
    def agregar_libro(self, libro):
        self.lista_libros.append(libro)
        
    def buscar(self, titulo):
        for lib in self.lista_libros:
            if lib.titulo == titulo:
                return lib
        return None
        
    def prestar(self, titulo, usuario):
        lib = self.buscar(titulo)
        if lib == None:
            print(f"No tenemos el libro {titulo}")
            return
            
        if lib.disponible:
            lib.disponible = False
            usuario.libros_prestados.append(lib)
            print(f"Se prestó {titulo} a {usuario.nombre}")
        else:
            print(f"El libro {titulo} ya está prestado")
            
    def devolver(self, titulo, usuario):
        for lib in usuario.libros_prestados:
            if lib.titulo == titulo:
                lib.disponible = True
                usuario.libros_prestados.remove(lib)
                print(f"{usuario.nombre} devolvió {titulo}")
                return
        print(f"{usuario.nombre} no tiene ese libro")
        
    def catalogo(self):
        print("--- Catálogo ---")
        for lib in self.lista_libros:
            estado = "Sí" if lib.disponible else "No"
            print(f"{lib.titulo} - Disponible: {estado}")
            
# Pruebas
biblio = Biblioteca()
biblio.agregar_libro(Libro_Biblio("POO Básico", "Pedro"))
biblio.agregar_libro(Libro_Biblio("Ingenieria de datos 1", "Maria"))

user1 = Usuario("Jose")

biblio.catalogo()
biblio.prestar("POO Básico", user1)
biblio.prestar("POO Básico", user1)
biblio.prestar("Libro falso", user1)
biblio.catalogo()
biblio.devolver("POO Básico", user1)
biblio.catalogo()
```
